```python
# Lets actually create the final project:
rates = {"USD" : 1,
         "NPR" : 150,
         "INR" : 88,
         "EUR" : 0.86}
# LETS STORE INFORMATION FROM USER:
amount = int(input("Enter the amount:"))
from_currency = input("Enter the from currency:").upper()
to_currency = input("Enter the to_currency:").upper()

if amount <= 0:
  print("Enter a valid amount: ")
elif from_currency not in rates or to_currency not in rates:
  print("Sorry that currency is not supported.")
else:
    USD_amount = amount / rates[from_currency]
    converted_amount = USD_amount * rates[to_currency]
    print("------------------CURRENCY CONVERTER------------------")
    print(f"{amount:.2f} {from_currency} = {converted_amount:.2f} {to_currency}")


```
