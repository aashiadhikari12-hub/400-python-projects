```python
def percentage_calculator():

    print("\n--- Percentage Calculator ---")

    try:
        obtained_marks = float(input("Enter obtained marks: "))
        total_marks = float(input("Enter total marks: "))

        if total_marks <= 0:
            print("Total marks must be greater than zero.")

        elif obtained_marks < 0:
            print("Obtained marks cannot be negative.")

        elif obtained_marks > total_marks:
            print("Obtained marks cannot exceed total marks.")

        else:
            percentage = (obtained_marks / total_marks) * 100

            print("Your percentage is:", round(percentage, 2), "%")

    except ValueError:
        print("Invalid input. Please enter numerical values.")


def cgpa_calculator():

    print("\n--- CGPA Calculator ---")

    try:
        total_subjects = int(input("Enter number of subjects: "))

        if total_subjects <= 0:
            print("Number of subjects must be greater than zero.")
            return

        total_weighted_points = 0
        total_credit_hours = 0

        for i in range(total_subjects):

            print(f"\nSubject {i + 1}")

            while True:
                try:
                    grade = float(input("Enter grade point (0-4): "))

                    if 0 <= grade <= 4:
                        break

                    print("Grade point must be between 0 and 4.")

                except ValueError:
                    print("Please enter a valid numerical grade.")

            while True:
                try:
                    credit = float(input("Enter credit hours: "))

                    if credit > 0:
                        break

                    print("Credit hours must be greater than zero.")

                except ValueError:
                    print("Please enter valid numerical credit hours.")

            total_weighted_points += grade * credit
            total_credit_hours += credit

        cgpa = total_weighted_points / total_credit_hours

        print("\nYour CGPA is:", round(cgpa, 2))

    except ValueError:
        print("Invalid input. Please enter a whole number of subjects.")


def main():

    while True:

        print("\n==============================")
        print(" Percentage and CGPA Calculator")
        print("==============================")

        print("1. Percentage Calculator")
        print("2. CGPA Calculator")
        print("3. Exit")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            percentage_calculator()

        elif choice == "2":
            cgpa_calculator()

        elif choice == "3":
            print("Thank you for using our calculator!")
            break

        else:
            print("Invalid choice. Please select 1, 2, or 3.")


main()
```
