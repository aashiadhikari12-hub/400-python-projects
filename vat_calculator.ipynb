```python
# we can also use function to get the value.
# Function to calculate VAT and final price
def get_price(price, rate):
    tax = price * (rate / 100)
    final_price = price + tax
    return tax, final_price


# Call the function outside the function definition
tax, final_price = get_price(9000, 13)
print(tax)
print(final_price)

print(get_price(80000, 13))
print(get_price(11000, 12))
print(get_price(13000, 13))

```

```python
# lets run all this together
def add_tax(price, rate):
    tax = price * rate / 100
    return tax


def extract_tax(total, rate):
    tax = total * rate / (100 + rate)
    return tax


def main():
    price = float(input("Enter price: NPR "))
    rate = float(input("Enter tax rate (%): "))

    mode = input("Type 'add' or 'extract': ").strip().lower()

    if price < 0 or rate < 0:
        print("Price and rate cannot be negative.")
        return

    if mode == "add":
        tax = add_tax(price, rate)
        total = price + tax
        original_price = price

    elif mode == "extract":
        tax = extract_tax(price, rate)
        total = price
        original_price = total - tax

    else:
        print("Invalid mode. Type add or extract.")
        return

    print(f"Original price: NPR {original_price:.2f}")
    print(f"Tax amount: NPR {tax:.2f}")
    print(f"Final price: NPR {total:.2f}")


if __name__ == "__main__":
    main()
```
