```python

# ==========================================
# PROJECT 9: PERSONAL EXPENSE TRACKER
# Beginner Python Project
# ==========================================

expenses = []
next_id = 1


# FEATURE 1: ADD EXPENSE

def add_expense():
    global next_id

    print("\n--- ADD EXPENSE ---")

    name = input("Expense name: ").strip()

    if not name:
        print("Expense name cannot be empty.")
        return

    try:
        amount = float(input("Amount (NPR): "))

        if amount <= 0:
            print("Amount must be greater than zero.")
            return

    except ValueError:
        print("Please enter a valid number.")
        return

    category = input("Category: ").strip()

    if not category:
        print("Category cannot be empty.")
        return

    expense = {
        "id": next_id,
        "name": name,
        "amount": amount,
        "category": category
    }

    expenses.append(expense)
    next_id += 1

    print("Expense added successfully!")


# FEATURE 2: VIEW ALL EXPENSES

def view_expenses():

    print("\n--- ALL EXPENSES ---")

    if not expenses:
        print("No expenses recorded.")
        return

    for expense in expenses:
        print(
            f"ID: {expense['id']} | "
            f"Name: {expense['name']} | "
            f"Amount: NPR {expense['amount']:.2f} | "
            f"Category: {expense['category']}"
        )


# FEATURE 3: SEARCH EXPENSE

def search_expense():

    print("\n--- SEARCH EXPENSE ---")

    search_name = input("Enter expense name: ").strip().lower()

    found = False

    for expense in expenses:

        if search_name in expense["name"].lower():

            print(expense)
            found = True

    if not found:
        print("No matching expense found.")


# FEATURE 4: EDIT EXPENSE

def edit_expense():

    print("\n--- EDIT EXPENSE ---")

    try:
        expense_id = int(input("Enter expense ID: "))
    except ValueError:
        print("Please enter a valid ID.")
        return

    for expense in expenses:

        if expense["id"] == expense_id:

            new_name = input("New name (Enter to keep current): ").strip()
            new_amount = input("New amount (Enter to keep current): ").strip()
            new_category = input("New category (Enter to keep current): ").strip()

            if new_name:
                expense["name"] = new_name

            if new_amount:
                try:
                    amount = float(new_amount)

                    if amount <= 0:
                        print("Amount must be greater than zero.")
                        return

                    expense["amount"] = amount

                except ValueError:
                    print("Invalid amount.")
                    return

            if new_category:
                expense["category"] = new_category

            print("Expense updated successfully!")
            return

    print("Expense ID not found.")


# FEATURE 5: DELETE EXPENSE

def delete_expense():

    print("\n--- DELETE EXPENSE ---")

    try:
        expense_id = int(input("Enter expense ID: "))
    except ValueError:
        print("Please enter a valid ID.")
        return

    for expense in expenses:

        if expense["id"] == expense_id:

            expenses.remove(expense)
            print("Expense deleted successfully!")
            return

    print("Expense ID not found.")


# FEATURE 6: TOTAL SPENDING

def show_total():

    total = sum(expense["amount"] for expense in expenses)

    print(f"\nTotal spending: NPR {total:.2f}")


# FEATURE 7: CATEGORY-WISE SPENDING

def category_report():

    print("\n--- CATEGORY REPORT ---")

    if not expenses:
        print("No expenses recorded.")
        return

    categories = {}

    for expense in expenses:

        category = expense["category"]

        if category not in categories:
            categories[category] = 0

        categories[category] += expense["amount"]

    for category, amount in categories.items():
        print(f"{category}: NPR {amount:.2f}")


# FEATURE 8: HIGHEST EXPENSE

def highest_expense():

    if not expenses:
        print("No expenses recorded.")
        return

    highest = max(expenses, key=lambda expense: expense["amount"])

    print("\n--- HIGHEST EXPENSE ---")
    print(f"Name: {highest['name']}")
    print(f"Amount: NPR {highest['amount']:.2f}")
    print(f"Category: {highest['category']}")


# MAIN APPLICATION

def main():

    while True:

        print("\n========== EXPENSE TRACKER ==========")
        print("1. Add Expense")
        print("2. View All Expenses")
        print("3. Search Expense")
        print("4. Edit Expense")
        print("5. Delete Expense")
        print("6. Total Spending")
        print("7. Category Report")
        print("8. Highest Expense")
        print("9. Exit")

        choice = input("\nChoose an option (1-9): ").strip()

        if choice == "1":
            add_expense()

        elif choice == "2":
            view_expenses()

        elif choice == "3":
            search_expense()

        elif choice == "4":
            edit_expense()

        elif choice == "5":
            delete_expense()

        elif choice == "6":
            show_total()

        elif choice == "7":
            category_report()

        elif choice == "8":
            highest_expense()

        elif choice == "9":
            print("Thank you for using Expense Tracker!")
            break

        else:
            print("Invalid option. Please try again.")


# RUN APPLICATION

main()
```
