```python
# lets make every password include all 4 character types:
# lets imprve by contanining atleast 1 loweracase, 1 uppercase, 1 digit, 1 symbol
import secrets
import string
lowercase = string.ascii_lowercase
uppercase = string.ascii_uppercase
digits = string.digits
symbols = "!@#$%&*"
all_characters = lowercase = uppercase + digits + symbols
length = int(input(" Enter the number of password thatis generated:"))
if length < 4:
  print("password length must be atleast 4")
else:
  password_characters = [secrets.choice(lowercase), secrets.choice(uppercase),
                         secrets.choice(digits), secrets.choice(symbols)]
for i in range(length - 4):
  password_characters.append(secrets.choice(all_characters))
  secrets.SystemRandom().shuffle(password_characters)
  password = "".join(password_characters)
  print(" Your password is:",password)
```
