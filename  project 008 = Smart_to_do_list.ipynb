```python
import json
import os
from datetime import date, datetime

# -------------------------------
# 1. DATA STORAGE
# -------------------------------

FILE_NAME = "todo_data.json"


def load_tasks():
    if os.path.exists(FILE_NAME):
        try:
            with open(FILE_NAME, "r") as file:
                return json.load(file)

        except (json.JSONDecodeError, OSError):
            print("Could not load saved tasks.")
            return []

    return []


def save_tasks():
    with open(FILE_NAME, "w") as file:
        json.dump(tasks, file, indent=4)


tasks = load_tasks()


# -------------------------------
# 2. ADD TASK
# -------------------------------

def add_task():

    title = input("Enter task name: ").strip()

    if not title:
        print("Task name cannot be empty.")
        return

    print("\nSelect Priority:")
    print("1. High")
    print("2. Medium")
    print("3. Low")

    priorities = {
        "1": "High",
        "2": "Medium",
        "3": "Low"
    }

    priority_choice = input("Choose priority: ")

    if priority_choice not in priorities:
        print("Invalid priority.")
        return

    due_date = input("Enter due date (YYYY-MM-DD), or press Enter to skip: ").strip()

    if due_date:
        try:
            datetime.strptime(due_date, "%Y-%m-%d")

        except ValueError:
            print("Invalid date format.")
            return

    new_task = {
        "title": title,
        "priority": priorities[priority_choice],
        "due_date": due_date,
        "completed": False
    }

    tasks.append(new_task)
    save_tasks()

    print("Task added successfully!")


# -------------------------------
# 3. VIEW TASKS
# -------------------------------

def view_tasks():

    if not tasks:
        print("No tasks available.")
        return

    print("\n========== YOUR TASKS ==========")

    for number, task in enumerate(tasks, start=1):

        status = "Completed" if task["completed"] else "Pending"

        due_date = task.get("due_date", "")
        due_text = due_date if due_date else "No due date"

        print(f"\nTask {number}: {task['title']}")
        print(f"Priority: {task.get('priority', 'Medium')}")
        print(f"Due date: {due_text}")
        print(f"Status: {status}")

        if due_date and not task["completed"]:
            if due_date < date.today().isoformat():
                print("WARNING: Overdue!")


# -------------------------------
# 4. COMPLETE TASK
# -------------------------------

def complete_task():

    if not tasks:
        print("No tasks available.")
        return

    view_tasks()

    try:
        choice = int(input("\nEnter task number to complete: "))

    except ValueError:
        print("Please enter a valid number.")
        return

    if choice < 1 or choice > len(tasks):
        print("Invalid task number.")
        return

    selected_task = tasks[choice - 1]

    if selected_task["completed"]:
        print("This task is already completed.")
        return

    selected_task["completed"] = True
    save_tasks()

    print("Task marked as completed!")


# -------------------------------
# 5. DELETE TASK
# -------------------------------

def delete_task():

    if not tasks:
        print("No tasks available.")
        return

    view_tasks()

    try:
        choice = int(input("\nEnter task number to delete: "))

    except ValueError:
        print("Please enter a valid number.")
        return

    if choice < 1 or choice > len(tasks):
        print("Invalid task number.")
        return

    removed_task = tasks[choice - 1]

    confirmation = input(
        f"Delete '{removed_task['title']}'? (y/n): "
    ).strip().lower()

    if confirmation == "y":
        tasks.pop(choice - 1)
        save_tasks()
        print("Task deleted successfully!")

    else:
        print("Deletion cancelled.")


# -------------------------------
# 6. PROGRESS DASHBOARD
# -------------------------------

def show_progress():

    total = len(tasks)

    completed = sum(1 for task in tasks if task["completed"])

    pending = total - completed

    percentage = (completed / total * 100) if total > 0 else 0

    print("\n========== TASK DASHBOARD ==========")
    print(f"Total tasks: {total}")
    print(f"Completed tasks: {completed}")
    print(f"Pending tasks: {pending}")
    print(f"Completion rate: {percentage:.1f}%")

    print("====================================")


# -------------------------------
# 7. SEARCH TASKS
# -------------------------------

def search_tasks():

    keyword = input("Enter task name to search: ").strip().lower()

    if not keyword:
        print("Please enter a search term.")
        return

    results = [
        task for task in tasks
        if keyword in task["title"].lower()
    ]

    if not results:
        print("No matching tasks found.")
        return

    print("\nSearch Results:")

    for task in results:
        status = "Completed" if task["completed"] else "Pending"
        print(f"- {task['title']} | {task.get('priority', 'Medium')} | {status}")


# -------------------------------
# 8. MAIN APPLICATION
# -------------------------------

def main():

    print("\nWelcome to Smart To-Do Manager!")

    while True:

        print("\n========== TO-DO MANAGER ==========")
        print("1. Add Task")
        print("2. View Tasks")
        print("3. Complete Task")
        print("4. Delete Task")
        print("5. View Progress")
        print("6. Search Tasks")
        print("7. Exit")

        choice = input("\nEnter your choice: ").strip()

        if choice == "1":
            add_task()

        elif choice == "2":
            view_tasks()

        elif choice == "3":
            complete_task()

        elif choice == "4":
            delete_task()

        elif choice == "5":
            show_progress()

        elif choice == "6":
            search_tasks()

        elif choice == "7":
            save_tasks()
            print("Your tasks have been saved. Goodbye!")
            break

        else:
            print("Invalid choice. Please select 1-7.")


# Start the application
main()
```
