```python

import random
import json
import os


# ==========================================
#           1. QUESTION BANK
# ==========================================

quiz_questions = [
    {
        "question": "Which keyword is used to define a function?",
        "options": ["A. function", "B. def", "C. create", "D. define"],
        "answer": "B"
    },
    {
        "question": "Which function displays output in Python?",
        "options": ["A. input()", "B. display()", "C. print()", "D. output()"],
        "answer": "C"
    },
    {
        "question": "Which data type stores True or False?",
        "options": ["A. String", "B. Integer", "C. Float", "D. Boolean"],
        "answer": "D"
    },
    {
        "question": "Which symbol is used for comments?",
        "options": ["A. //", "B. #", "C. /*", "D. @"],
        "answer": "B"
    },
    {
        "question": "Which function converts a value into an integer?",
        "options": ["A. str()", "B. float()", "C. int()", "D. bool()"],
        "answer": "C"
    }
]


# ==========================================
#           2. WELCOME FUNCTION
# ==========================================

def welcome_user(name):
    print("\nWelcome", name)
    print("Let's test your Python knowledge!")


# ==========================================
#           3. DISPLAY QUESTION
# ==========================================

def display_question(number, item):
    print("\n------------------------------")
    print("Question", number)
    print(item["question"])

    for option in item["options"]:
        print(option)


# ==========================================
#           4. CHECK ANSWER
# ==========================================

def check_answer(user_answer, correct_answer):
    return user_answer == correct_answer


# ==========================================
#           5. CALCULATE PERCENTAGE
# ==========================================

def calculate_percentage(score, total):
    return (score / total) * 100


# ==========================================
#           6. INPUT VALIDATION
# ==========================================

def get_valid_answer():

    while True:
        answer = input("Your answer (A/B/C/D): ").strip().upper()

        if answer in ["A", "B", "C", "D"]:
            return answer

        print("Invalid input! Enter A, B, C or D.")


# ==========================================
#           7. REPLAY FUNCTION
# ==========================================

def play_again():

    while True:
        choice = input("\nPlay again? (yes/no): ").strip().lower()

        if choice == "yes":
            return True

        elif choice == "no":
            return False

        else:
            print("Invalid choice! Enter yes or no.")


# ==========================================
#           8. SAVE RESULTS USING JSON
# ==========================================

def save_result(result):

    filename = "quiz_results.json"

    if os.path.exists(filename):

        try:
            with open(filename, "r") as file:
                results = json.load(file)

            if not isinstance(results, list):
                results = []

        except (json.JSONDecodeError, OSError):
            results = []

    else:
        results = []

    results.append(result)

    with open(filename, "w") as file:
        json.dump(results, file, indent=4)

    print("\nYour result has been saved!")


# ==========================================
#           9. VIEW QUIZ HISTORY
# ==========================================

def view_history():

    filename = "quiz_results.json"

    if not os.path.exists(filename):
        print("\nNo previous results found.")
        return

    try:
        with open(filename, "r") as file:
            results = json.load(file)

    except (json.JSONDecodeError, OSError):
        print("\nUnable to read quiz history.")
        return

    if not isinstance(results, list) or not results:
        print("\nNo previous results found.")
        return

    print("\n========== QUIZ HISTORY ==========")

    for number, result in enumerate(results, start=1):

        print("\nAttempt:", number)
        print("Name:", result["name"])
        print("Score:", result["score"], "/", result["total"])
        print("Percentage:", result["percentage"], "%")

    print("\n==================================")


# ==========================================
#           10. MAIN PROGRAM
# ==========================================

print("================================")
print("       PYTHON KNOWLEDGE TEST")
print("================================")

user_name = input("Enter your name: ").strip()

while not user_name:
    print("Name cannot be empty.")
    user_name = input("Enter your name: ").strip()

welcome_user(user_name)


# ==========================================
#           11. QUIZ LOOP
# ==========================================

while True:

    score = 0

    # Select 3 random questions
    selected_questions = random.sample(quiz_questions, 3)

    # Ask questions
    for number, item in enumerate(selected_questions, start=1):

        display_question(number, item)

        user_answer = get_valid_answer()

        if check_answer(user_answer, item["answer"]):
            print("Correct answer!")
            score += 1

        else:
            print("Incorrect answer!")
            print("Correct answer:", item["answer"])

    # Calculate result
    total_questions = len(selected_questions)

    percentage = calculate_percentage(score, total_questions)

    # Display result
    print("\n========== QUIZ RESULT ==========")
    print("Participant:", user_name)
    print("Total questions:", total_questions)
    print("Correct answers:", score)
    print("Wrong answers:", total_questions - score)
    print("Percentage:", round(percentage, 2), "%")

    if percentage == 100:
        print("Excellent! Perfect score!")

    elif percentage >= 60:
        print("Great job! Keep improving!")

    else:
        print("Keep practicing!")

    # Prepare result
    result = {
        "name": user_name,
        "score": score,
        "total": total_questions,
        "percentage": round(percentage, 2)
    }

    # Save result
    save_result(result)

    print("================================")

    # Replay
    if not play_again():
        break


# ==========================================
#           12. SHOW QUIZ HISTORY
# ==========================================

view_history()

print("\nThank you for playing!")
print("See you next time!")
```

```python

# =====================================================
#       PYTHON KNOWLEDGE QUIZ - COMPLETE PROJECT
#                  ONE-CELL SETUP
# =====================================================

import os
import json
import zipfile

# Create project folder
PROJECT = "python-knowledge-quiz"
os.makedirs(PROJECT, exist_ok=True)


# =====================================================
# 1. STREAMLIT APPLICATION
# =====================================================

app_code = r'''
import streamlit as st
import random
import json
import os

st.set_page_config(
    page_title="Python Knowledge Quiz",
    page_icon="🐍",
    layout="centered"
)

# ---------------- QUESTION BANK ----------------

quiz_questions = [
    {
        "question": "Which keyword defines a function in Python?",
        "options": ["A. function", "B. def", "C. create", "D. define"],
        "answer": "B"
    },
    {
        "question": "Which function displays output?",
        "options": ["A. input()", "B. display()", "C. print()", "D. output()"],
        "answer": "C"
    },
    {
        "question": "Which data type stores True or False?",
        "options": ["A. String", "B. Integer", "C. Float", "D. Boolean"],
        "answer": "D"
    },
    {
        "question": "Which symbol is used for comments?",
        "options": ["A. //", "B. #", "C. /*", "D. @"],
        "answer": "B"
    },
    {
        "question": "Which function converts a value into an integer?",
        "options": ["A. str()", "B. float()", "C. int()", "D. bool()"],
        "answer": "C"
    }
]

RESULT_FILE = "quiz_results.json"


# ---------------- SAVE RESULTS ----------------

def save_result(result):
    if os.path.exists(RESULT_FILE):
        try:
            with open(RESULT_FILE, "r") as file:
                results = json.load(file)
            if not isinstance(results, list):
                results = []
        except (json.JSONDecodeError, OSError):
            results = []
    else:
        results = []

    results.append(result)

    with open(RESULT_FILE, "w") as file:
        json.dump(results, file, indent=4)


# ---------------- SESSION STATE ----------------

defaults = {
    "started": False,
    "finished": False,
    "current": 0,
    "score": 0,
    "answered": False,
    "saved": False,
    "questions": [],
    "feedback": "",
    "correct": False,
    "player_name": ""
}

for key, value in defaults.items():
    if key not in st.session_state:
        st.session_state[key] = value


# ---------------- START NEW QUIZ ----------------

def start_quiz(name):
    st.session_state.player_name = name
    st.session_state.questions = random.sample(
        quiz_questions, len(quiz_questions)
    )
    st.session_state.current = 0
    st.session_state.score = 0
    st.session_state.started = True
    st.session_state.finished = False
    st.session_state.answered = False
    st.session_state.saved = False


# ---------------- PAGE DESIGN ----------------

st.title("🐍 Python Knowledge Quiz")

st.write(
    "How well do you know Python? "
    "Take this short challenge and find out!"
)

st.divider()


# ---------------- WELCOME SCREEN ----------------

if not st.session_state.started:

    st.subheader("Welcome, future Python developer!")

    name = st.text_input(
        "What should we call you?",
        placeholder="Enter your name"
    )

    if st.button("Start Quiz", type="primary"):
        if name.strip():
            start_quiz(name.strip())
            st.rerun()
        else:
            st.warning("Please enter your name.")


# ---------------- QUIZ SCREEN ----------------

elif not st.session_state.finished:

    current = st.session_state.current
    total = len(st.session_state.questions)
    question = st.session_state.questions[current]

    st.subheader(f"Welcome, {st.session_state.player_name}!")

    st.progress(current / total)

    st.caption(f"Question {current + 1} of {total}")

    st.markdown(f"### {question['question']}")

    if not st.session_state.answered:

        selected = st.radio(
            "Choose your answer:",
            question["options"],
            key=f"answer_{current}"
        )

        if st.button("Submit Answer", type="primary"):

            selected_letter = selected[0]

            if selected_letter == question["answer"]:
                st.session_state.score += 1
                st.session_state.feedback = "Correct! Well done."
                st.session_state.correct = True
            else:
                st.session_state.feedback = (
                    "Not quite! The correct answer is "
                    + question["answer"]
                )
                st.session_state.correct = False

            st.session_state.answered = True
            st.rerun()

    else:

        if st.session_state.correct:
            st.success(st.session_state.feedback)
        else:
            st.error(st.session_state.feedback)

        if st.button("Next Question", type="primary"):

            st.session_state.current += 1
            st.session_state.answered = False

            if st.session_state.current >= total:
                st.session_state.finished = True

            st.rerun()


# ---------------- RESULTS SCREEN ----------------

if st.session_state.finished:

    total = len(st.session_state.questions)
    score = st.session_state.score
    percentage = (score / total) * 100

    if not st.session_state.saved:

        result = {
            "name": st.session_state.player_name,
            "score": score,
            "total": total,
            "percentage": round(percentage, 2)
        }

        save_result(result)
        st.session_state.saved = True

    st.balloons()

    st.subheader("🎉 Quiz Completed!")

    st.write(f"Well done, {st.session_state.player_name}!")

    col1, col2 = st.columns(2)

    col1.metric("Your Score", f"{score}/{total}")
    col2.metric("Percentage", f"{percentage:.2f}%")

    if percentage == 100:
        st.success("Perfect score! Excellent work!")
    elif percentage >= 60:
        st.success("Great job! Keep improving.")
    else:
        st.info("Keep practicing. You can do it!")

    with st.expander("View your answer review"):
        for i, question in enumerate(st.session_state.questions):
            st.write(f"**Question {i + 1}:** {question['question']}")
            st.write(f"Correct answer: {question['answer']}")

    if st.button("Play Again", type="primary"):
        start_quiz(st.session_state.player_name)
        st.rerun()


# ---------------- FOOTER ----------------

st.divider()
st.caption("Built with Python and Streamlit | Python Knowledge Quiz")
'''

with open(f"{PROJECT}/app.py", "w") as file:
    file.write(app_code)


# =====================================================
# 2. REQUIREMENTS FILE
# =====================================================

with open(f"{PROJECT}/requirements.txt", "w") as file:
    file.write("streamlit\n")


# =====================================================
# 3. README FILE
# =====================================================

readme = """
# Python Knowledge Quiz

An interactive Python quiz web application built using Python and Streamlit.

## Features
- Randomized multiple-choice questions
- Player name input
- Instant answer feedback
- Score and percentage calculation
- Answer review
- Replay functionality
- JSON result saving

## Technologies
- Python
- Streamlit
- JSON

## Run locally

Install dependencies:

    pip install -r requirements.txt

Run the application:

    streamlit run app.py

## Project purpose

Created as a hands-on Python learning project to practice
functions, dictionaries, lists, loops, conditionals, randomization,
file handling and JSON.
"""

with open(f"{PROJECT}/README.md", "w") as file:
    file.write(readme)


# =====================================================
# 4. INITIAL RESULTS FILE
# =====================================================

with open(f"{PROJECT}/quiz_results.json", "w") as file:
    json.dump([], file, indent=4)


# =====================================================
# 5. PACKAGE ALL PROJECT FILES
# =====================================================

zip_name = "python-knowledge-quiz.zip"

with zipfile.ZipFile(zip_name, "w") as zip_file:
    for filename in os.listdir(PROJECT):
        zip_file.write(
            os.path.join(PROJECT, filename),
            arcname=f"{PROJECT}/{filename}"
        )


# =====================================================
# 6. DISPLAY PROJECT CONTENTS
# =====================================================

print("==========================================")
print("       PROJECT CREATED SUCCESSFULLY")
print("==========================================")

print("\nProject files:")

for filename in os.listdir(PROJECT):
    print("✔", filename)

print("\nZIP package:", zip_name)
print("\nYour project is ready for testing and GitHub!")
```
