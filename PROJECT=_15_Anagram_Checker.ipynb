```python
#==================================
# PROJECT ANAGRAM CHECKER
# DESCRIPTION: CHECKS WHETHER TWO WORDS CONTAONS THE SAME LETTER:
#===================================
def is_anagram(word1, word2):
  word1 = word1.lower().replace(" ", "")
  word2 = word2.lower().replace(" ", "")

  if not word1 or not word2:
      return None
  return sorted(word1) == sorted(word2)

def main():
  print("WELCOME TO THE ANAGRAM CHECKER.")
  word1 = input("Enter the first word.")
  word2 = input("Enter the second word.")

  result = is_anagram(word1, word2)

  if result is None:
    print("Please provide both words.")
  elif result:
    print("Those words are anagram.")
  else:
    print("Those words are not anagrams.")

if __name__ == "__main__":
  main()
```
