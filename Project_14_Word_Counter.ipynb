```python
# I will be learning more word counter useful by calculationg characers without spaces, avaerage word length and number of sentences.
text = "Python is fun to learn. Coding is interesting. "
characters_without_spaces = len(text.replace(" ", ""))
print(f"Characters without spaces: {characters_without_spaces}")
sentences = text.count(".") + text.count("!") + text.count("?")
average_word_length = characters_without_spaces / len(text)
print(average_word_length)
print(sentences)
```
